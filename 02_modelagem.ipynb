```python
import pandas as pd
```

Modelando o dataset para exportação


```python
# importando o dataset limpo
df = pd.read_csv('../data/processed/dataset_v1.csv')
```

```python
dim_exame = df[
    [
        'CodigoMatrix',
        'CodigoFaturamento',
        'NomeProc',
        'Material',
        'Meio'
    ]
].drop_duplicates()
```

```python
dim_area = df[
    [
        'Area',
        'NomeArea',
        'UnidadeProdutiva'
    ]
].drop_duplicates()
```

```python
dim_origem = df[
    [
        'GrupoOrigem',
        'NomeGrupoOrigem',
        'Origem',
        'NomeOrigem',
        'Clinica',
        'NomeClinica'
    ]
].drop_duplicates()
```

```python
dim_paciente = df[
    [
        'Sexo'
    ]
].drop_duplicates()
```

```python
dim_area = dim_area.reset_index(drop=True)

dim_area['IDArea'] = dim_area.index + 1
```

```python
dim_exame = dim_exame.reset_index(drop=True)

dim_exame['IDExame'] = dim_exame.index + 1
```

```python
dim_origem = dim_origem.reset_index(drop=True)

dim_origem['IDOrigem'] = dim_origem.index + 1
```

```python
dim_paciente = dim_paciente.reset_index(drop=True)

dim_paciente['IDPaciente'] = dim_paciente.index + 1
```

```python
df = df.merge(
    dim_exame,
    on=[
        'CodigoMatrix',
        'CodigoFaturamento',
        'NomeProc',
        'Material',
        'Meio'
    ],
    how='left'
)
```

```python
df = df.merge(
    dim_area,
    on=[
        'Area',
        'NomeArea',
        'UnidadeProdutiva'
    ],
    how='left'
)
```

```python
df = df.merge(
    dim_origem,
    on=[
        'GrupoOrigem',
        'NomeGrupoOrigem',
        'Origem',
        'NomeOrigem',
        'Clinica',
        'NomeClinica'
    ],
    how='left'
)
```

```python
df = df.merge(
    dim_paciente,
    on=['Sexo'],
    how='left'
)
```

```python
fato_exames = df[
    [
        'NumeroPedido',

        'IDExame',
        'IDArea',
        'IDOrigem',
        'IDPaciente',

        'DataSistema',
        'DataHoraPedido',
        'DataCriacaoAmostra',
        'HoraCriacaoAmostra',
        'DataHoraImpressao',
        'DataHoraColeta',
        'DataHoraTriagem',
        'DataHoraResultado',
        'DataHoraLiberacaoTecnica',
        'DataHoraLiberacaoClinica',
        'DataHoraCancelamento',
        'DataHoraRetiradaProc',

        'Recoleta',
        'ResultadoRetificado',
        'FlagLiberacaoClinicaAutomatica'
    ]
]
```
